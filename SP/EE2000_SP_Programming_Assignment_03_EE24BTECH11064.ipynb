<dt> <h4> 4. Fast Fourier Transform (FFT) algorithm :  <dt> <h4>

Objective : Computing the N-point DFT of a signal requires $N^{2}$ complex multiplications and additions. However, the Fast Fourier Transform (FFT) algorithm optimizes the DFT algorithm by exploiting its redundant operations, resulting in an efficient computation of the N-point DFT algorithm. Specifically, the FFT algorithm computes the N-point DFT of the signal in $O(Nlog_{2}N)$ time complexity. While this reduction may seem trivial for small N, it is crucial for large values of N that arise in many practical applications. For example, a speech signal sampled at 8kHz would have 8000 samples per second, and computing the N-point DFT through the basic approach would require $N^{2} = 64$ million complex multiplications and additions. In contrast, the FFT algorithm only requires $Nlog_{2}N = 0.10$ million operations. This demonstrates the importance of the FFT algorithm in efficiently computing the DFT algorithm.
```python
#All imports
import numpy as np

#######################################################
#Part(a) : (1)
#######################################################
#Write a code to compute the N-point DFT of the sequence
def DFT(x,N):
    """
    #Inputs : x is input sequence, and N is the number of points for which DTFT is evaluated
    #Outputs : The N-point DTFT of the signal
    #if N > length of the input sequence, pad the input sequence with zeros to get N-point signal
    #if N < legnth of the input sequence, truncate the input sequence to obtain the N-point signal
    """
    ####################################################
    #Pad / Truncate the input sequence to obtain the sequence of length N
    x = np.array(x, dtype=np.complex128)
    L = len(x)
    if L < N:
        x = np.pad(x, (0, N - L))
    elif L > N:
        x = x[:N]

    ####################################################


    ####################################################
    #Compute DFT
    X = np.zeros_like(x, dtype=np.complex128)  #Initialize with zeros, make sure that dtype is complex, as we may get complex numbers in DFT
    for k in range(0,N,1):
        for n in range(0,N,1):
          X[k] += x[n] * np.exp(-1j * 2 * np.pi * k * n / N)

    ####################################################
    #Return the N-point DFT
    return X
```

```python
#All imports
import numpy as np

#######################################################
#Part(a) : (2)
#######################################################
#Write a code to compute the N-point signal from its N-point DFT using the Inverse Discrete Fourier Transform (IDFT) operation
def IDFT(X,N):
    """
    #Inputs : X is the N-point DFT of the sequence x[n]
    #Outputs : The N-point signal x[n]
    """
    ####################################################
    #Compute DFT
    x = np.zeros_like(X, dtype=np.complex128)
    for n in range(0,N,1):
        for k in range(0,N,1):
            x[n] += X[k] * np.exp(1j * 2 * np.pi * k * n / N)
        x[n] /= N

    ####################################################
    #Return the N-point time domain sequence
    return x
```

```python
#All imports
import numpy as np

#######################################################
#Part(a) : (3)
#######################################################
#Matrix-vector formulation of DFT & IDFT
def DFT_Matrix(x,N):
    """
    #Inputs : x is input sequence, and N is the number of points for which DTFT is evaluated
    #Outputs : The N-point DTFT of the signal
    #if N > length of the input sequence, pad the input sequence with zeros to get N-point signal
    #if N < legnth of the input sequence, truncate the input sequence to obtain the N-point signal
    """
    ####################################################
    #Pad / Truncate the input sequence to obtain the sequence of length N
    x = np.array(x, dtype=np.complex128)
    L = len(x)
    if L < N:
        x = np.pad(x, (0, N - L))
    elif L > N:
        x = x[:N]

    ####################################################
    ####################################################
    #Create the DFT matrix
    n = np.arange(N)
    k = n.reshape((N, 1))
    W = np.exp(-1j * 2 * np.pi * k * n / N)

    ####################################################
    ####################################################
    #Compute DFT
    #Multiply DFT maxtrix with input sequence to obtain N-Point DFT of the signal

    X = np.dot(W, x)

    ####################################################
    #Return the N-point DFT
    return X

def IDFT_Matrix(X,N):
    """
    #Inputs : X is the N-point DFT of the sequence x[n]
    #Outputs : The N-point signal x[n]
    """
    ####################################################
    #Create the IDFT matrix
    n = np.arange(N)
    k = n.reshape((N, 1))
    W_I = np.exp(1j * 2 * np.pi * k * n / N) / N


    ####################################################
    ####################################################
    #Compute the time domain signal
    #Multiply IDFT maxtrix with the DFT sequence to obtain the time domain signal
    x = np.dot(W_I, X)

    ####################################################
    #Return the N-point time domain sequence
    return x
```

```python
#All imports
import numpy as np
```
```python
import time

#######################################################
#Part(b) : (6)  Different ways of computing the linear convolution
#######################################################

#######################################################
#Create a function to compute the linear convolution of the sequences x and h
#######################################################
def linear_convolution_direct(x, h):
    """
    Inputs : Two sequences : x,h
    Outputs : Linear convolution of the two sequences computed through the direct implementation of the convolution formula
    """
    x = np.array(x, dtype=float)
    h = np.array(h, dtype=float)
    #Find the length of the input sequences
    N = len(x)
    M = len(h)
    #Initialize an array with zeros to store the final result
    L = N + M - 1
    #Compute the linear convolution of the two sequence x and h
    y = np.zeros(L, dtype=float)
    for n in range(L):
        for k in range(N):
            if 0 <= n - k < M:
                y[n] += x[k] * h[n - k]
    #Return the linear convolution of the two sequences x and h
    return y


#######################################################
#Compute circular convolution of two sequences (Direct)
#######################################################
def circular_convolution_direct(x, h):
    """
    Inputs : Two sequences : x,h
    Outputs : Circular convolution computed directly using formula
    """
    N = len(x)
    y = np.zeros(N, dtype=complex)
    for n in range(N):
        for k in range(N):
            y[n] += x[k] * h[(n - k) % N]
    return y


#######################################################
#Compute circular convolution using DFT property
#######################################################
def circular_convolution_from_property(x, h):
    """
    Inputs : Two sequences : x,h
    Outputs : Circular convolution computed using DFT property
    """
    X = np.fft.fft(x)
    H = np.fft.fft(h)
    Y = X * H
    y = np.fft.ifft(Y).real  # Take real part (imaginary part due to rounding)
    return y


#######################################################
#Compute linear convolution of two sequences using circular convolution
#######################################################
def linear_convolution_from_circular_convolution_direct(x, h):
    """
    Inputs : Two sequences : x,h
    Outputs : Linear convolution computed using circular convolution (direct formula)
    """
    M = len(x)
    N = len(h)
    L = M + N - 1
    x_pad = np.pad(x, (0, L - M))
    h_pad = np.pad(h, (0, L - N))
    y = circular_convolution_direct(x_pad, h_pad)
    return y


#######################################################
#Compute linear convolution of two sequences using DFT property
#######################################################
def linear_convolution_from_circular_convolution_property(x, h):
    """
    Inputs : Two sequences : x,h
    Outputs : Linear convolution computed using DFT property
    """
    M = len(x)
    N = len(h)
    L = M + N - 1
    x_pad = np.pad(x, (0, L - M))
    h_pad = np.pad(h, (0, L - N))
    y = circular_convolution_from_property(x_pad, h_pad)
    return y


#######################################################
#Compare the computation complexity involved
#######################################################
N = 2**10
x = np.random.randn(N)
M = 2**8
h = np.random.randn(M)

#--- Linear Convolution (Direct)
start_time = time.time()
Y1 = linear_convolution_direct(x, h)
end_time = time.time()
time_direct = end_time - start_time

#--- Linear Convolution via Circular Convolution (Direct)
start_time = time.time()
Y2 = linear_convolution_from_circular_convolution_direct(x, h)
end_time = time.time()
time_circular_direct = end_time - start_time

#--- Linear Convolution via Circular Convolution (Using DFT Property)
start_time = time.time()
Y3 = linear_convolution_from_circular_convolution_property(x, h)
end_time = time.time()
time_circular_property = end_time - start_time

#######################################################
#Display results
#######################################################
print("Computation Time for Different Linear Convolution Approaches")
print("1. Direct Linear Convolution:                                ", time_direct)
print("2. Linear Convolution via Circular Convolution (Direct):      ", time_circular_direct)
print("3. Linear Convolution via Circular Convolution (DFT Property):", time_circular_property)
```

```python
#All imports
import numpy as np
```
```python
import matplotlib.pyplot as plt

#######################################################
#Part(c) : (2)
#######################################################

#######################################################
#Creat a function to perform low-pass filtering of the signal with a cut-off frequency of fc
#######################################################
def lowpass_filter(x,fc):
    """
    Inputs : Input sequence and the cut-off frequency of the low-pass filter
    Outputs : Low-pass filtered input sequence
    """
    #Compute the DFT of the input sequence
    N = len(x)
    # Compute the DFT of the input sequence
    X = np.fft.fft(x)
    # Frequency axis
    freqs = np.fft.fftfreq(N, d=1/fs)

    #Set zeros in the DFT sequence corresponding to the frequencies greater than the cut-off frequency.
    X[np.abs(freqs) > fc] = 0

    #Perform IDFT operation on the modfied DFT sequence
    y = np.fft.ifft(X)

    #Return the low-pass filtered signal
    return y

#######################################################
#Low pass filtering on synthetic
#######################################################

#Generate 10 Hz sinusoidal with a sampling frequency of 2000 Hz
# Sampling frequency
fs = 2000
t = np.arange(0, 0.2, 1/fs)
x1 = np.sin(2 * np.pi * 10 * t)
#Generate 100 Hz sinusoidal with a sampling frequency of 2000 Hz
x2 = np.sin(2 * np.pi * 100 * t)

#Combine 10 Hz and 1000 Hz sinusoidal signals

x_combined = x1 + x2
#Pass the added signal through a low-pass filter with a cut-off frequency of 50 Hz

fc = 50
y_filtered = lowpass_filter(x_combined, fc)


#Plot the original added signal and the filtered in signal subplots
SMALL_SIZE = 20
MEDIUM_SIZE = 20
BIGGER_SIZE = 20
plt.rc('font', size=SMALL_SIZE)
plt.rc('axes', titlesize=SMALL_SIZE)
plt.rc('axes', labelsize=MEDIUM_SIZE)
plt.rc('xtick', labelsize=SMALL_SIZE)
plt.rc('ytick', labelsize=SMALL_SIZE)
plt.rc('legend', fontsize=SMALL_SIZE)
plt.rc('figure', titlesize=BIGGER_SIZE)

plt.figure(figsize=(20,10))
# Original signal
plt.stem(t, x_combined, 'b')
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Original Combined Signal (10 Hz + 1000 Hz)")
plt.grid(True)

# Filtered signal
plt.figure(figsize=(20,10))
plt.stem(t, y_filtered, 'r')
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title(f"Low-pass Filtered Signal (Cut-off = {fc} Hz)")
plt.grid(True)

plt.tight_layout()
plt.show()
#Report your observations




```

```python
#All imports
import soundfile as sf
from google.colab import drive
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

#######################################################
#Part(c) : (3)
#######################################################
#Simulating dereverberation

drive.mount('/content/drive')

#Loading the previous signals required for dereverberation
sig, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/speech.wav")
rir_smallroom, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/rir_smallroom.wav")
rir_mediumroom, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/rir_mediumroom.wav")
rir_largeroom, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/rir_largeroom.wav")


reverbe_small, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/speech_reverbe_small.wav")
reverbe_medium, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/speech_reverbe_medium.wav")
reverbe_large, samplerate = sf.read("/content/drive/MyDrive/SP-PA3/speech_reverbe_large.wav")

def derev(signal, rir):
    N = len(signal) + len(rir) - 1
    Xr = np.fft.fft(signal, N)
    H = np.fft.fft(rir, N)
    H_inv = np.zeros_like(H, dtype=complex)

    # Inversion (While avoiding division by zero)
    H_inv[H != 0] = 1 / H[H != 0]

    # Apply inverse filtering
    X_est = Xr * H_inv
    x_hat = np.fft.ifft(X_est)
    return np.real(x_hat)

dereverb_small = derev(reverbe_small, rir_smallroom)
dereverb_medium = derev(reverbe_medium, rir_mediumroom)
dereverb_large = derev(reverbe_large, rir_largeroom)

print("Playing Original Speech:")
display(Audio(sig, rate=samplerate))

print("Playing Reverberated Speech (Small Room):")
display(Audio(reverbe_small, rate=samplerate))
print("Dereverberated Speech (Small Room):")
display(Audio(dereverb_small, rate=samplerate))

print("Playing Reverberated Speech (Medium Room):")
display(Audio(reverbe_medium, rate=samplerate))
print("Dereverberated Speech (Medium Room):")
display(Audio(dereverb_medium, rate=samplerate))

print("Playing Reverberated Speech (Large Room):")
display(Audio(reverbe_large, rate=samplerate))
print("Playing Dereverberated Speech (Large Room):")
display(Audio(dereverb_large, rate=samplerate))

#Report your observations
```
